# 2
영문이름을 입력하면 웹사이트 주소가 리턴되는 함수, get_url() 를  구현하세요.
예) url = get_url('google') 
print(url) # www.google.com

```python
name = input('웹사이트 이름: ')

def get_url(name):
    url = 'www.' + name + '.com'
    return url

print(get_url(name))
```

# 3
구구단을 수행할 단번호를 입력하면 그 단의 9까지의 결과를 나열하는 프로그램을 입력하세요.
예) 3 --> 3 6 9 12 15 18 21 24 27

```python
n = int(input('구구단 단번호 입력: '))
for i in range(1,10):
    print(n * i, end=" ")
```
